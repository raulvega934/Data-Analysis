```python
import pandas as pd
import matplotlib.pyplot as plt

jobs_skills = pd.read_csv("/Users/raulvegasoto/Desktop/Data Analysis/project_sql/Top_paying_jobs_skills.csv")
skill_pay = pd.read_csv("/Users/raulvegasoto/Desktop/Data Analysis/project_sql/average_salary_skills.csv")

# 1. Top paying jobs
jobs = jobs_skills.drop_duplicates("job_id").sort_values("annual_salary")
jobs.plot.barh(x="company_name", y="annual_salary", legend=False, figsize=(9, 5))
plt.title("Top Paying Remote Data Analyst Jobs")
plt.xlabel("Annual salary ($)")
plt.ylabel("")
plt.tight_layout()
plt.show()

# 2. Skills in those jobs
skills = jobs_skills["skills"].value_counts().head(10).sort_values()
skills.plot.barh(figsize=(9, 5))
plt.title("Most Common Skills in the Top Paying Jobs")
plt.xlabel("Number of top jobs requiring the skill")
plt.ylabel("")
plt.tight_layout()
plt.show()

# 3. Average salary by skill
skill_pay = skill_pay.sort_values("average_salary")
skill_pay.plot.barh(x="skills", y="average_salary", legend=False, figsize=(9, 7))
plt.title("Average Salary by Skill")
plt.xlabel("Average yearly salary ($)")
plt.ylabel("")
plt.tight_layout()
plt.show()
```

```python
demand = pd.read_csv("/Users/raulvegasoto/Desktop/Data Analysis/project_sql/skils_demand.csv").sort_values("skill_occurence")
demand.plot.barh(x="skills", y="skill_occurence", legend=False, figsize=(9, 5))
plt.title("Most In-Demand Skills for Data Analysts")
plt.xlabel("Number of postings")
plt.ylabel("")
plt.tight_layout()
plt.show()
```
